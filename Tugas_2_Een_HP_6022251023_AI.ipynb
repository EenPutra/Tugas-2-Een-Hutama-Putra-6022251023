```python
# --- FUZZY LOGIC TEMPERATURE MEMBERSHIP FUNCTIONS ---
import numpy as np
import matplotlib.pyplot as plt

# --------------------------
# MEMBERSHIP FUNCTIONS
# --------------------------

def dingin(t):
    if t <= 15:
        return 1
    elif 15 < t <= 25:
        return (25 - t) / 10
    else:
        return 0

# SEDANG: puncak di 25°C
def sedang(t):
    if 15 < t <= 25:
        return (t - 15) / 10      # naik ke 1 di suhu 25
    elif 25 < t <= 35:
        return (35 - t) / 10      # turun dari 1 ke 0
    else:
        return 0

def panas(t):
    if t <= 25:
        return 0
    elif 25 < t <= 35:
        return (t - 25) / 10
    else:
        return 1

# --------------------------
# PLOT MEMBERSHIP FUNCTIONS
# --------------------------

T = np.linspace(10, 45, 1000)
μ_dingin = [dingin(x) for x in T]
μ_sedang = [sedang(x) for x in T]
μ_panas = [panas(x) for x in T]

plt.figure(figsize=(10,6))

plt.plot(T, μ_dingin, 'r', linewidth=2, label="Dingin")
plt.plot(T, μ_sedang, 'g', linewidth=2, label="Sedang (puncak 25°C)")
plt.plot(T, μ_panas, 'b', linewidth=2, label="Panas")

plt.title("Fungsi Keanggotaan Suhu (Fuzzy Logic)")
plt.xlabel("Temperatur (°C)")
plt.ylabel("Derajat Keanggotaan μ")
plt.grid(True)
plt.legend()
plt.show()


```

```python
# --------------------------
# INPUT SUHU
# --------------------------

t_input = float(input("Masukkan suhu (°C): "))

μ_d = dingin(t_input)*100
μ_s = sedang(t_input)*100
μ_p = panas(t_input)*100

print("\n=== HASIL FUZZIFIKASI ===")
print(f"Suhu {t_input}°C:")
print(f"  → Dingin : {μ_d:.1f} %")
print(f"  → Sedang : {μ_s:.1f} %")
print(f"  → Panas  : {μ_p:.1f} %")

# Menentukan kategori dominan
kategori = max(
    ("Dingin", μ_d),
    ("Sedang", μ_s),
    ("Panas", μ_p),
    key=lambda x: x[1]
)

print(f"\nKlasifikasi fuzzy: **{kategori[0]}**")
```
